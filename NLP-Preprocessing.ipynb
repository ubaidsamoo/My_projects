# NLP Preprocessing Example

This are common NLP preprocessing steps:
- Lowercasing
- Removing HTML tags
- Removing numbers
- Removing punctuation
- Removing emojis
- Removing stopwords
- Removing extra spaces


```python
# Original text
text = """
<h1>sooo yesterday!!!</h1> i taught the <b>students</b> about <i>nlp preprocessing</i> ,yeah yeah like lower  CASE , stop words, and punctuation ... oh wow!!what a  mess mess... 
there were about 21 or 22 students 999 888 maybe ??not sure 🤔 anyway we talked and talked and talked about commas, dots...???"""
print(text)
```

## Step 1: Convert text to lowercase


```python
text = text.lower()
print(text)
```

## Step 2: Remove HTML tags


```python
import re

text = re.sub(r'<.*?>', ' ', text)
print(text)
```

## Step 3: Remove numbers


```python
text = re.sub(r'\d+', ' ', text)
print(text)
```

## Step 4: Remove punctuation and emojis


```python
text = re.sub(r'[^\w\s]', ' ', text)
print(text)
```

## Step 5: Remove stopwords

```python
import re

# manual stopword list
stop_words = {
    "the","a","an","is","of","in","to","on","for","and","that","by",
    "we","were","about","like","oh","yeah","anyway","this","it","as",
    "at","be","from","or","are","but","not"
}

# preprocessing
text = text.lower()
text = re.sub(r'<.*?>', ' ', text)       # remove HTML
text = re.sub(r'\d+', ' ', text)         # remove numbers
text = re.sub(r'[^\w\s]', ' ', text)     # remove punctuation/emojis
text = re.sub(r'\s+', ' ', text).strip() # remove extra spaces

words = text.split()
text = " ".join(w for w in words if w not in stop_words)

print(text)
```

## Step 6: Remove extra spaces (final clean text)


```python
text = re.sub(r'\s+', ' ', text).strip()
print(text)
```
